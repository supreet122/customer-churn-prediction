```python
import pandas as pd 
import numpy as np 

import matplotlib.pyplot as plt 
import seaborn as sns 



#Better-looking graph
plt.style.use("ggplot") 
```

```python
df = pd.read_csv("../data/telco_clean.csv")
df.head()
```

```python
print(df.shape)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Churn Label",data=df)

plt.title("Customer Churn Distribution ")
plt.show()
```

```python
plt.Figure(figsize=(6,4))
sns.countplot(x="Gender",hue="Churn Label",data=df)
plt.title("Gender vs Churn")
plt.show()
```

```python
plt.Figure(figsize=(6,4))
sns.countplot(x="Senior Citizen",hue="Churn Label",data=df)
plt.title("Senior Citizen vs Churn")
plt.show()
```

```python
plt.Figure(figsize=(10,5))
sns.countplot(x="Offer",hue="Churn Label",data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Internet Type",hue="Churn Label",data=df)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Monthly Charge"],bins=30)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Churn Label", y="Monthly Charge", data=df)

plt.show()
```

```python
plt.Figure(figsize=(8,5))
sns.histplot(df["Tenure in Months"],bins=30)
plt.show()
```

```python
plt.Figure(figsize=(8,5))
sns.boxplot(
    x="Churn Label",
    y="Tenure in Months",
    data=df
)
plt.show()
```

```python
plt.figure(figsize=(15,10))
numeric_df = df.select_dtypes(include=np.number)
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)
plt.show()
```

```python
## Business Insights

#Most customers do not churn.

#Customers with shorter tenure are more likely to churn.

#Higher monthly charges are associated with higher churn.

#Certain internet service types show higher churn rates.

#Senior citizens appear to churn at a higher rate.

#Some offers are associated with better customer retention.
```
